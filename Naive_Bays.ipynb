```python
# Bibliotheken importieren
import pandas as pd
```

## Load data for model-training
    source data link: https://www.kaggle.com/datasets/saurabh00007/iriscsv 

```python
df = pd.read_csv('./data/Iris.csv')
df.drop('Id', axis=1, inplace=True) # axis = 0 -> Zeile, axis = 1 -> Spalte
x = df.drop('Species', axis=1)
print(x.columns)
y = df['Species']
```

## training model - GaussianNB

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# random_state=42, damit die Daten(random) aber immer gleich aufgeteilt werden, damit die Ergebnisse reproduzierbar sind
# x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# shuffle=False, damit die Daten nicht gemischt werden, da die Daten bereits sortiert sind
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, shuffle=False)

print(f'Trainingsdaten: {x_train.shape}')
# print(f'Traindaten: {x_train.columns}')
model = GaussianNB()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)

accuracy = accuracy_score(y_test, y_pred) 
print(f'Accuracy: {accuracy:.2f}')
```
