```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
# load the dataset
df = pd.read_csv('./data/Iris.csv')
df.drop(columns=['Id'], inplace=True) # inplace = True means that the changes will be made to the original dataframe
x = df.drop(columns=['Species'])
y = df['Species']
```

```python
# Data Splitting, training the model
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
model = LogisticRegression(max_iter=200) # max_iter is the maximum number of iterations for the solver to converge
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
# print(y_pred)
```

```python
df_kundenzufriedenheit = pd.DataFrame({
    'Anzahl der Kellner':[2,5,8],
    'Wartezeit(Min.)':[8,7,6],
    'Kundenzufriedenheit':[50,75,95]
    })
df_add_more_training_data = pd.DataFrame({
    'Anzahl der Kellner':[3,6,9],
    'Wartezeit(Min.)':[7,6,5],
    'Kundenzufriedenheit':[60,80,100]}
    )
df_kundenzufriedenheit = pd.concat([df_kundenzufriedenheit, df_add_more_training_data], ignore_index=True)
print(f'{df_kundenzufriedenheit}\n')

x_kundenzufriedenheit = df_kundenzufriedenheit.drop(columns=['Kundenzufriedenheit'])
y_kundenzufriedenheit = df_kundenzufriedenheit['Kundenzufriedenheit']

x_train_kunden, x_test_kunden, y_train_kunden, y_test_kunden = train_test_split(x_kundenzufriedenheit, y_kundenzufriedenheit, test_size=0.2)
model_kundenzufriedenheit = LogisticRegression(max_iter=1)
print(f"x_train_kunden:\n {x_train_kunden}, \n y_train_kunden: \n{y_train_kunden}")
model_kundenzufriedenheit.fit(x_train_kunden, y_train_kunden)
y_pred_kundenzufriedenheit = model_kundenzufriedenheit.predict(x_test_kunden)
print(f'y_pred_kundenzufriedenheit: {y_pred_kundenzufriedenheit}\n')

print(f'accuracy: {accuracy_score(y_test_kunden, y_pred_kundenzufriedenheit)}')
print(classification_report(y_test_kunden, y_pred_kundenzufriedenheit))
```
