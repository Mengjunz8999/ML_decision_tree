```python
import math
import matplotlib.pyplot as plt
```

```python
#----------------------------------------------------
#1. Berechnen Sie die Distanz jedes Punktes zu beiden Zentren. 
#----------------------------------------------------
orginal_x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12] 
orginial_y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21] 
z1_x = 7
z1_y  = 20
z2_x = 8
z2_y = 25

fig = plt.figure()
plt.scatter(orginal_x, orginial_y, color='blue')
plt.scatter(z1_x, z1_y, color='red',s=100)
plt.annotate('Z₁(7,20)', xy=(7, 20), xytext=(6.5, 19.2), fontsize=10, color='red') # annotate(): Text hinzufügen
plt.scatter(8, 25, color='green', label='Zentrum 2',s=100)
plt.annotate('Z₂(8,25)', xy=(8, 25), xytext=(7.5, 24.3), fontsize=10, color='green')
plt.xlabel('x')
plt.ylabel('y')

Distanz_round1 = []
Distanz_round2 = []

def cul_distanze(x_point_list,y_point_list,z_x_point,z_y_point,Distanz_list):
    for i,j in zip(x_point_list,y_point_list):
        distanze_x = (z_x_point - i) ** 2
        distanze_y = (z_y_point - j) ** 2
        distanze = distanze_x + distanze_y
        distanze = math.sqrt(distanze)
        distanze = round(distanze, 2)
        Distanz_list.append(distanze)
    return distanze

cul_distanze(orginal_x, orginial_y, z1_x, z1_y,Distanz_round1)
cul_distanze(orginal_x, orginial_y, z2_x, z2_y,Distanz_round2)
   
print(f'Distanz_round1: {Distanz_round1}') 
print(f'Distanz_round2: {Distanz_round2}')


# Cluster zuordnen
cluster1_x, cluster1_y = [], []
cluster2_x, cluster2_y = [], []

for i in range(len(orginal_x)):
    if Distanz_round1[i] <= Distanz_round2[i]:
        cluster1_x.append(orginal_x[i])
        cluster1_y.append(orginial_y[i])
    else:
        cluster2_x.append(orginal_x[i])
        cluster2_y.append(orginial_y[i])

print(f'Cluster 1: {list(zip(cluster1_x, cluster1_y))}')
print(f'Cluster 2: {list(zip(cluster2_x, cluster2_y))}')

# Plot aktualisieren
fig_zuorden = plt.figure()
plt.scatter(cluster1_x, cluster1_y, color='blue',   s=80, label='Cluster 1')
plt.scatter(cluster2_x, cluster2_y, color='orange', s=80, label='Cluster 2')

plt.scatter(z1_x, z1_y, color='red',   s=150)
plt.annotate('Z₁(7,20)', xy=(z1_x, z1_y), xytext=(6.5, 19.2), fontsize=10, color='red')
plt.scatter(z2_x, z2_y, color='green', s=150)
plt.annotate('Z₂(8,25)', xy=(z2_x, z2_y), xytext=(7.5, 24.3), fontsize=10, color='green')

plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc = 'upper left') 

plt.show()

#----------------------------------------------------
#2. Berechnen Sie die neuen Zentren der Cluster.
#----------------------------------------------------

Clustering1_x = [4, 5,  4, 3,  6, 10, 12] 
Clustering1_y = [21, 19, 17, 16, 22, 21, 21] 
summe1_x = 0
summe1_y = 0

for i,j in zip(Clustering1_x,Clustering1_y):
    summe1_x += i
    summe1_y += j
new_z1_x = summe1_x / len(Clustering1_x)
new_z1_x = round(new_z1_x, 2)
new_z1_y = summe1_y / len(Clustering1_y)
new_z1_y = round(new_z1_y, 2)
print(f'summe_x: {summe1_x}, summe_y: {summe1_y}')
print(f'Neues Zentrum Z₁: ({new_z1_x}, {new_z1_y})')

Clustering2_x = [10,11,14] 
Clustering2_y = [24,25,24] 
summe2_x = 0
summe2_y = 0
for i,j in zip(Clustering2_x,Clustering2_y):
    summe2_x += i
    summe2_y += j
new_z2_x = summe2_x / len(Clustering2_x)
new_z2_x = round(new_z2_x, 2)
new_z2_y = summe2_y / len(Clustering2_y)
new_z2_y = round(new_z2_y, 2)
print(f'summe_x: {summe2_x}, summe_y: {summe2_y}')
print(f'Neues Zentrum Z₁: ({new_z2_x}, {new_z2_y})')

#----------------------------------------------------
#new zentern zuordnen und plotten
#----------------------------------------------------
new_cluster1_x, new_cluster1_y = [], []
new_cluster2_x, new_cluster2_y = [], []

# i don't acturlly need the list, because 
distanze_list1 = []
distanze_list2 = []

for i in range(len(orginal_x)): # recalulate distance to new centers
    new1_distanze_x = (new_z1_x - orginal_x[i]) ** 2
    new1_distanze_y = (new_z1_y - orginial_y[i]) ** 2
    new_distanze1 = new1_distanze_x + new1_distanze_y
    new_distanze1 = math.sqrt(new_distanze1)
    new_distanze1 = round(new_distanze1, 2)
    distanze_list1.append(new_distanze1)

    new2_distanze_x = (new_z2_x - orginal_x[i]) ** 2
    new2_distanze_y = (new_z2_y - orginial_y[i]) ** 2
    new_distanze2 = new2_distanze_x + new2_distanze_y
    new_distanze2 = math.sqrt(new_distanze2)
    new_distanze2 = round(new_distanze2, 2)
    distanze_list2.append(new_distanze2)

    if new_distanze1 <= new_distanze2:
        new_cluster1_x.append(orginal_x[i])
        new_cluster1_y.append(orginial_y[i])
    else:
        new_cluster2_x.append(orginal_x[i])
        new_cluster2_y.append(orginial_y[i])

print(f'New Cluster 1: {list(zip(new_cluster1_x, new_cluster1_y))}')
print(f'New Cluster 2: {list(zip(cluster2_x, cluster2_y))}')
print(f'distanze_list1: {distanze_list1}, distanze_list2: {distanze_list2}')

# Plot aktualisieren
new_fig_zuorden = plt.figure()
plt.scatter(new_cluster1_x, new_cluster1_y, color='blue',   s=80, label='Cluster 1')
plt.scatter(new_cluster2_x, new_cluster2_y, color='orange', s=80, label='Cluster 2')

plt.scatter(new_z1_x, new_z1_y, color='red',   s=150)
plt.annotate(f'Z₁({new_z1_x},{new_z1_y})', xy=(new_z1_x, new_z1_y), xytext=(new_z1_x, new_z1_y + 0.5), fontsize=10, color='red')
plt.scatter(new_z2_x, new_z2_y, color='green', s=150)
plt.annotate(f'Z₂({new_z2_x},{new_z2_y})', xy=(new_z2_x, new_z2_y), xytext=(new_z2_x, new_z2_y +0.5), fontsize=10, color='green')

plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc = 'upper left') 

plt.show()
```
