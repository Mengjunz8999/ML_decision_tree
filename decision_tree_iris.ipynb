## Schritt 1: Bibliotheken importieren

**Erklärung:**  
Hier importieren wir alle Bibliotheken, die wir brauchen. `pandas` verwenden wir zum Laden und Bearbeiten der CSV-Datei. `matplotlib` brauchen wir für die Visualisierung des Entscheidungsbaums. Von `sklearn` importieren wir das Decision-Tree-Modell, die Funktion zum Aufteilen der Daten und Bewertungsmethoden.



````python
# Bibliotheken importieren

import pandas as pd  
# pandas: Wird verwendet, um Daten zu laden und zu bearbeiten (z. B. CSV-Dateien)

import matplotlib.pyplot as plt  
# matplotlib: Bibliothek zum Erstellen von Diagrammen und Visualisierungen


from sklearn.tree import DecisionTreeClassifier, plot_tree  
# DecisionTreeClassifier: Modell für Entscheidungsbäume (Klassifikation)
# plot_tree: Funktion, um den Entscheidungsbaum grafisch darzustellen

from sklearn.model_selection import train_test_split  
# train_test_split: Teilt Daten in Trainings- und Testdaten auf

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix  
# accuracy_score: Berechnet die Genauigkeit des Modells
# classification_report: Liefert Precision, Recall, F1-Score (detaillierte Bewertung)
# confusion_matrix: Zeigt, wie viele Vorhersagen korrekt/falsch pro Klasse sind
````

## Schritt 2: CSV-Datei laden

**Erklärung:**  
In diesem Schritt laden wir die Datei `Iris.csv` in ein DataFrame. Ein DataFrame ist eine tabellarische Datenstruktur von pandas. Die Datei muss im gleichen Ordner wie das Notebook liegen, damit sie direkt mit `pd.read_csv()` geladen werden kann.

**Aufgabe:**  
Laden Sie die CSV-Datei `Iris.csv` in die Variable `df`.

**Lösung:**


````python
df = pd.read_csv("./data/Iris.csv", index_col=False)
print(f" --------------Iris data: -------------------- \n {df}")
````

## Schritt 3: Erste Zeilen anzeigen

**Erklärung:**  
Mit `df.head()` schauen wir uns die ersten fünf Zeilen der Tabelle an. Dadurch können wir prüfen, ob die Datei richtig geladen wurde und wie die Daten ungefähr aussehen.

**Aufgabe:**  
Zeigen Sie die ersten Zeilen des DataFrames an.

**Lösung:**


````python
df.head()
````

## Schritt 4: Spalten prüfen

**Erklärung:**  
Hier lassen wir uns alle Spaltennamen anzeigen. Das ist wichtig, damit wir wissen, welche Spalten als Eingabedaten verwendet werden können und welche Spalte die Zielvariable ist.

**Aufgabe:**  
Geben Sie alle Spaltennamen des DataFrames aus.

**Lösung:**


````python
colums_list = df.columns
print(f"-------------colums_list------------------\n{colums_list}")
````

## Schritt 5: Id-Spalte entfernen

**Erklärung:**  
Falls eine Spalte `Id` vorhanden ist, entfernen wir sie. Diese Spalte ist nur eine Nummerierung und enthält keine sinnvolle Information für die Klassifikation der Iris-Blumen.

**Aufgabe:**  
Entfernen Sie die Spalte `Id`, falls sie im DataFrame vorhanden ist.

**Lösung:**


````python
def df_drop_col(df, *col)-> df.DataFrame:
    df_drop_col = df.drop(columns=list(col))
    return df_drop_col

df_drop_id = df_drop_col(df, 'Id')

print(f"------------------df without id column---------------------- \n {df_drop_id.columns}")
````

## Schritt 6: Features und Zielvariable trennen

**Erklärung:**  
Jetzt trennen wir die Eingabedaten von der Zielvariable. `X` enthält die Merkmale der Blume, also zum Beispiel Länge und Breite von Kelch- und Blütenblättern. `y` enthält die Klasse, also die Iris-Art.

**Aufgabe:**  
Erstellen Sie `X` mit allen Merkmalen und `y` mit der Zielspalte `Species`.

**Lösung:**


````python
X = df_drop_col(df,'Id','Species')
print(X)
y = df['Species']
print(y)
````

## Schritt 7: Daten in Trainings- und Testdaten aufteilen

**Erklärung:**  
Die Daten werden in Trainingsdaten und Testdaten aufgeteilt. Mit den Trainingsdaten lernt das Modell. Mit den Testdaten prüfen wir später, wie gut das Modell neue, unbekannte Daten klassifizieren kann. Hier verwenden wir 80 Prozent Training und 20 Prozent Test.



````python
# Daten in Trainings- und Testdaten aufteilen

# X = Eingabedaten (Features)
# y = Zielwerte (Klassen)

# Wir teilen die Daten in zwei Teile:
# - Trainingsdaten → zum Lernen des Modells
# - Testdaten → zum Überprüfen, wie gut das Modell ist
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    
    X,  # Eingabedaten (z. B. SepalLength, PetalWidth, ...)
    
    y,  # Zielvariable (z. B. Iris-Art)
    
    test_size=0.2,  # 20% der Daten werden für den Test verwendet
                    # 80% bleiben fürs Training
    
    random_state=42 # sorgt dafür, dass die Aufteilung immer gleich ist
)
````

## Schritt 8: Decision Tree Modell erstellen

**Erklärung:**  
Hier erstellen wir das Modell. Ein Decision Tree trifft Entscheidungen anhand von Regeln. Mit `max_depth=3` begrenzen wir die Tiefe des Baums. Dadurch wird das Modell einfacher und Overfitting kann reduziert werden.

**Aufgabe:**  
### Aufgabe

Erstelle einen Decision Tree Classifier mit folgenden Parametern:

- `max_depth=3` (maximale Tiefe des Baums)
- `random_state=42` (für reproduzierbare Ergebnisse)
- `min_samples_leaf=10` (mindestens 10 Datenpunkte pro Blatt)

Speicheren Sie das Modell in einer Variable namens `model`.



**Lösung:**


````python
model = DecisionTreeClassifier(
    max_depth=3,
    random_state= 42,
    min_samples_leaf= 10
)
````

## Schritt 9: Modell trainieren

**Erklärung:**  
Jetzt trainieren wir das Modell mit den Trainingsdaten. Das Modell lernt dabei Zusammenhänge zwischen den Merkmalen in `X_train` und den Klassen in `y_train`.

**Aufgabe:**  
Trainieren Sie das Modell mit den Trainingsdaten.

**Lösung:**


````python
model.fit(X_train,y_train)
````

## Schritt 10: Vorhersagen auf Testdaten machen

**Erklärung:**  
Nach dem Training lassen wir das Modell Vorhersagen für die Testdaten machen. Die Ergebnisse speichern wir in `y_pred`. Diese Vorhersagen vergleichen wir später mit den echten Klassen aus `y_test`.

**Aufgabe:**  
Erstellen Sie Vorhersagen für die Testdaten und speichere sie in `y_pred`.

**Lösung:**


````python
y_pred = model.predict(X_test)
print(type(y_test))
````

````python
# Wir erstellen eine neue Tabelle (DataFrame) zum Vergleichen
vergleich = pd.DataFrame({
    
    # Spalte mit den echten Klassen aus den Testdaten
    # .values wandelt die pandas Series in ein Array um
    "Echte Klasse": y_test,
    
    # Spalte mit den Vorhersagen des Modells
    "Vorhersage": y_pred
})

# Zeigt die ersten 5 Zeilen der Vergleichstabelle an
vergleich.head()
````

## Schritt 11: Genauigkeit berechnen

**Erklärung:**  
Die Accuracy zeigt, wie viele Testbeispiele richtig klassifiziert wurden. Ein Wert von `1.0` bedeutet 100 Prozent richtig, ein Wert von `0.9` bedeutet 90 Prozent richtig.

**Aufgabe:**  
Berechnen Sie die Accuracy des Modells und geben Sie sie aus.

Verwenden Sie dafür die Funktion `accuracy_score` aus `sklearn.metrics`.
Vergleichen Sie die echten Werte (`y_test`) mit den Vorhersagen (`y_pred`).

**Lösung:**


````python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print(f"Genauigkeit des Modells: {accuracy:.2f}")

print("Training set size:", X_train.shape)
print("Test set size:", X_test.shape)
print("Training accuracy:", model.score(X_train, y_train))
print("Test accuracy:", model.score(X_test, y_test))
````

## Schritt 12: Decision Tree visualisieren

**Erklärung:**  
Jetzt visualisieren wir den trainierten Entscheidungsbaum. Dadurch kann man sehen, welche Regeln der Baum gelernt hat und nach welchen Merkmalen er Entscheidungen trifft.


````python
# Decision Tree visualisieren

# Erstellt eine neue Grafik mit bestimmter Größe (Breite=14, Höhe=8)
plt.figure(figsize=(14, 10))

# Zeichnet den Decision Tree
plot_tree(
    model,                     # Das trainierte Modell, das visualisiert wird
    
    feature_names=X.columns,   # Namen der Eingabe-Features (z. B. SepalLength, ...)
    
    class_names=model.classes_,# Namen der Klassen (z. B. Setosa, Versicolor, ...)
    
    filled=True               # Färbt die Knoten je nach Klasse (bessere Übersicht)
)

# Zeigt die Grafik an
plt.show()
````

## Schritt 13: Einzelne neue Blume vorhersagen

**Erklärung:**  
Zum Schluss testen wir das Modell mit einer neuen Blume. Die Werte müssen in der gleichen Reihenfolge angegeben werden wie die Spalten in `X`: `SepalLengthCm`, `SepalWidthCm`, `PetalLengthCm`, `PetalWidthCm`.

**Aufgabe:**  
Erstellen Sie eine neue Blume mit den Werten `[5.1, 3.5, 1.4, 0.2]` und lasse die Klasse vorhersagen.

**Lösung:**


````python
self_test = [[5.1, 3.5, 1.4, 0.2],[6.0,2.2,3.9,1.0]]
self_test_pred = model.predict(self_test) # if self_test is 1D array, then parameter is [self_test]
print(f"Vorhersage für Testdaten 0 : {self_test_pred[0]}")
print(f"Vorhersage für Testdaten 1 : {self_test_pred[1]}")
````
